# Ensemble Kalman Filter: State & Parameter Estimation <a class="anchor" id="enKF"></a>

Real-time data assimilation combines a numerical model with noisy observations to estimate the most likely state of a system. We use the **Ensemble Kalman Filter (EnKF)** — a sequential method that assimilates data *on the fly*, without storing or post-processing the full data history.

---


The EnKF fuses two sources of information:

1. **Observations** $\boldsymbol{d} \sim \mathcal{N}(\boldsymbol{d}^\dagger, \boldsymbol{\mathsf{C}}_{dd})$ — unbiased but noisy sensor measurements
2. **Model forecast** — a dynamical system propagating the state $\boldsymbol{\phi}$ forward in time

$$\mathrm{d}\boldsymbol{\phi} = \mathcal{F}(\boldsymbol{\phi} + \boldsymbol{\epsilon}_\phi,\, \boldsymbol{\alpha} + \boldsymbol{\epsilon}_\alpha)\,\mathrm{d}t$$

such that the model prediction on the observations is $\mathcal{M}(\boldsymbol{\phi}) \sim \mathcal{N}(\boldsymbol{d}^\dagger, \boldsymbol{\mathsf{C}}_{\phi\phi})$, where $\mathcal{M}$ is the measurement operator mapping the state variables to the measurement space.

Since both sources are uncertain, we model all errors as Gaussian and seek the **maximum a posteriori** (MAP) estimate — the state $\boldsymbol{\phi}^a$ that minimises:

$$\mathcal{J}(\boldsymbol{\phi}) = \underbrace{\|\boldsymbol{\phi} - \boldsymbol{\phi}^f\|^2_{(\boldsymbol{\mathsf{C}}_{\phi\phi}^f)^{-1}}}_{\text{model penalty}} + \underbrace{\|\boldsymbol{d} - \mathcal{M}(\boldsymbol{\phi})\|^2_{\boldsymbol{\mathsf{C}}_{dd}^{-1}}}_{\text{observation penalty}}$$


To estimate both state $\boldsymbol{\phi}$ and parameters $\boldsymbol{\alpha}$ simultaneously, we define an **augmented state vector** $\boldsymbol{\psi} = [\boldsymbol{\phi};\, \boldsymbol{\alpha}; \mathcal{M}(\boldsymbol{\phi})]$, treating parameters as time-constant state variables:

$$\mathrm{d}\begin{bmatrix}\boldsymbol{\phi} \\ \boldsymbol{\alpha}\end{bmatrix} = \begin{bmatrix}\mathcal{F}(\boldsymbol{\phi} + \boldsymbol{\epsilon}_\phi,\, \boldsymbol{\alpha} + \boldsymbol{\epsilon}_\alpha) \\ \boldsymbol{0}\end{bmatrix}\mathrm{d}t$$

with this, the cost function becomes

$$\mathcal{J}(\boldsymbol{\psi}) = \underbrace{\|\boldsymbol{\psi} - \boldsymbol{\psi}^f\|^2_{(\boldsymbol{\mathsf{C}}_{\psi\psi}^f)^{-1}}}_{\text{model penalty}} + \underbrace{\|\boldsymbol{d} - \boldsymbol{\mathsf{M}}\boldsymbol{\psi}\|^2_{\boldsymbol{\mathsf{C}}_{dd}^{-1}}}_{\text{observation penalty}}$$


---

## The EnKF Update <a class="anchor" id="update"></a>

The EnKF propagates $m$ ensemble members $\boldsymbol{\psi}_j$ to approximate the forecast statistics (mean and covariance)

$$\bar{\boldsymbol{\psi}} = \frac{1}{m}\sum_{j=1}^m \boldsymbol{\psi}_j, \qquad \boldsymbol{\mathsf{C}}_{\psi\psi} \approx \frac{1}{m-1}\sum_{j=1}^m (\boldsymbol{\psi}_j - \bar{\boldsymbol{\psi}})(\boldsymbol{\psi}_j - \bar{\boldsymbol{\psi}})^\top$$

When observations $\boldsymbol{d}$ arrive, each member is updated as:

$$\boldsymbol{\psi}_j^\mathrm{a} = \boldsymbol{\psi}_j^\mathrm{f} + \boldsymbol{\mathsf{K}}\left[\boldsymbol{d}_j - \boldsymbol{\mathsf{M}}\boldsymbol{\psi}_j^\mathrm{f}\right]$$

where the **Kalman gain** $\boldsymbol{\mathsf{K}}$ optimally weights model and observation uncertainty:

$$\boldsymbol{\mathsf{K}} = \boldsymbol{\mathsf{C}}_{\psi\psi}^\mathrm{f}\,\boldsymbol{\mathsf{M}}^\top\!\left(\boldsymbol{\mathsf{C}}_{dd} + \boldsymbol{\mathsf{M}}\boldsymbol{\mathsf{C}}_{\psi\psi}^\mathrm{f}\boldsymbol{\mathsf{M}}^\top\right)^{-1}$$

## Test case: twin experiment on the Lorenz63 model <a class="anchor" id="twin"></a>

We validate the EnKF on the **Lorenz 63** system — a canonical chaotic model:

$$\frac{\mathrm{d}x}{\mathrm{d}t} = \sigma(y - x), \qquad \frac{\mathrm{d}y}{\mathrm{d}t} = x(\rho - z) - y, \qquad \frac{\mathrm{d}z}{\mathrm{d}t} = xy - \beta z$$

with state $\boldsymbol{\phi} = [x;\,y;\,z]$ and parameters $\boldsymbol{\alpha} = [\sigma;\,\rho;\,\beta]$.

This system is **chaotic** at $\boldsymbol{\alpha} = [10;\,28;\,8/3]$: two trajectories initialised 0.01% apart diverge exponentially, with a predictability horizon set by the Lyapunov time $T_\lambda = 1/\lambda_\mathrm{max} \approx 1/0.91$. This makes it an ideal test for data assimilation — the EnKF must continuously correct the ensemble to track the true trajectory before it diverges.

We use a **twin experiment**: a synthetic truth $\boldsymbol{\psi}^\dagger$ is integrated from the model, observations are generated as $\boldsymbol{d} = \boldsymbol{\mathsf{M}}\boldsymbol{\psi}^\dagger + \boldsymbol{\epsilon}_d$, and the EnKF attempts to recover both $\boldsymbol{\phi}^\dagger$ and $\boldsymbol{\alpha}^\dagger$ without access to the truth.

### 1) The true state and parameters <a class="anchor" id="truth"></a>



```python

from romda.models.physical import Lorenz63
import numpy as np
rng = np.random.default_rng(0)

dt_t = 0.015
t_lyap = 0.9056 ** (-1)  # Lyapunov Time (inverse of maximal Lyapunov exponent)

true_params = dict(dt=dt_t,
                   rho=28.,
                   sigma=10.,
                   beta=8. / 3.,
                   psi0=rng.random(3)+10,
                   observe_dims=[0, 2]     # Select the dimensions to observe
                   )

# Initialize model
true_case = Lorenz63(**true_params)

# Forecast model
t_max = t_lyap * 100
psi, t = true_case.time_integrate(int(t_max / true_case.dt))
true_case.update_history(psi, t)
```

### 2) The observations <a class="anchor" id="obs"></a>

```python
from romda.observations import Observations

# # Draw data points from raw data
dt_obs = t_lyap * .5  # time between analyses
Nt_obs = dt_obs // dt_t
t_start, t_stop = t_lyap * 10, t_lyap * 60 # start and end of assimilation


truth = Observations(model=true_case, 
                    t_start=t_start, 
                    t_stop=t_stop, 
                    Nt_obs=Nt_obs, 
                    add_noise=True,
                    noise_type='gauss, add',
                    noise_level=0.05,
                    )

Observations.plot_truth(truth)
```

### 3) Define the ensemble case <a class="anchor" id="ens"></a>
We create an ensemble with randomly generated states and parameters

```python
from romda.estimators import EnSRKF, EnKF

alpha0 = dict(rho=(25., 35.),
              beta=(2, 4),
              sigma=(5,15))

ensemble = EnSRKF(parent_model=Lorenz63,      
                    dt=dt_t,             
                    m=50,               # Number of ensemble members
                    std_phi=0.2,        # Initial uncertainty in the state
                    std_alpha=alpha0,       # Initial uncertainty in the parameters
                    observe_dims=true_case.observe_dims
                    )
```

```python
# Forecast the ensemble without assimilation and visualize
ensemble.forecast_step(t_end=t_start-2*t_lyap, close=True)
ensemble.visualize_state(time_indices=[0,-1])
```

### 4) Apply data assimilation <a class="anchor" id="da"></a>
We now have all the ingredients to start our data assimilation algorithm.

```python

filter_ens = ensemble.copy()
filter_ens.inflation_factor = 1.02
# Measurement error covariance matrix: the covariance of the noise that `Observations` adds to the truth,
# i.e. a standard deviation of noise_level * max|y| for each observable
from romda.observations import observation_covariance
Cdd = observation_covariance(truth, std_obs=truth.noise_level)

# ----------------------------------------------------------------------------

for d, t_d in zip(truth.y_obs, truth.t_obs):
    filter_ens.forecast_step(t_end=t_d)
    filter_ens.analysis_step(d=d, Cdd=Cdd)
    
# ----------------------------------------------------------------------------

#Forecast the ensemble further without assimilation
t_extra = filter_ens.model.hist_t[-1] + filter_ens.model.t_CR
filter_ens.forecast_step(t_extra, averaged=False, close=True)
```

```python
print("number of assimilation steps: len(truth.t_obs) =", len(truth.t_obs))
```

```python
filter_ens.visualize_history(plot_members=True, 
                             truth=truth, 
                             reference_t=t_lyap,
                             reference_a=true_case.alpha0)
```

```python
# Visualize attractors
# from plot_results import plot_attractor
import matplotlib.pyplot as plt

case0 = true_case.copy()

filter_ens.model.close()
case1 = filter_ens.model.copy()

# Forecast both cases
Nt = 25 * int(t_lyap / case0.dt)
psi0, t0 = case0.time_integrate(Nt=Nt)
psi_ens, t1 = case1.time_integrate(Nt=Nt)

psi1 = psi_ens[:, :3, 0]
psi2 = psi_ens[:, :3, -1]

case0.visualize_attractor(psi_cases=[psi0, psi1, psi2])
plt.legend(['True system', 'Ensemble j = 0', 'Ensemble j = m'], frameon=False);
```

## Sampling frequency against the Lyapunov time <a class="anchor" id="sampling"></a>

In a chaotic system, an error grows on average as $\mathrm{e}^{t/T_\lambda}$, where $T_\lambda$ is the Lyapunov time. The filter corrects only the error that has grown since the last analysis, so the time between observations, $\Delta t_\mathrm{obs}$, must be compared with $T_\lambda$. The helpers of this section are in `aux_1X.py`.

```python
from aux_1X import (OI_BLUE, OI_GREY, error_scale, plot_inflation, plot_ks_fields, plot_ks_sweeps, plot_sampling,
                    plot_spread_between_analyses, plot_spread_growth, plot_tracking, sweep, twin_ks, twin_lorenz63,
                    use_tutorial_style)

use_tutorial_style()
attractor = np.asarray(true_case.hist)[:, :, 0]
case15 = Lorenz63(rho=15., dt=dt_t, psi0=attractor[-1])
psi15, _ = case15.time_integrate(int(5 * t_lyap / dt_t))       # on the attractor of rho = 15 (a fixed point)
plot_spread_growth([(28., attractor[-1], OI_BLUE, r'$\rho = 28$ (chaotic)'),
                    (15., psi15[-1, :, 0], OI_GREY, r'$\rho = 15$ (fixed point)')], error_scale(attractor))
```

The spread of a free ensemble grows at the rate of the leading Lyapunov exponent when the system is chaotic, and decays when it is not. Twin experiments with $x$ and $z$ observed and the parameters known:

```python
plot_tracking({rf'$\Delta t_\mathrm{{obs}} = {dt}\,T_\lambda$': twin_lorenz63(dt) for dt in (0.1, 0.5, 2)})

dt_obs_values = [0.05, 0.1, 0.25, 0.5, 1., 1.5, 2.]
sampling = sweep(twin_lorenz63, dt_obs_values, score=None)
plot_sampling(dt_obs_values, sampling)
print('median error:', np.round(np.median(sampling[:, :, 0], axis=1), 2))
```

The filter tracks the state when $\Delta t_\mathrm{obs} \lesssim 0.25\,T_\lambda$, and has almost no skill from $\Delta t_\mathrm{obs} \approx T_\lambda$, when the ensemble spreads over the attractor between analyses. The sampling frequency must be several times the leading Lyapunov exponent.

## Is inflation necessary if the system is chaotic? <a class="anchor" id="inflation"></a>

A finite ensemble underestimates its spread, and the multiplicative inflation (`inflation_factor`, $\rho_I > 1$) compensates for it. In a chaotic system, the dynamics inflate the spread between analyses:

```python
plot_spread_between_analyses(twin_lorenz63(0.25))

factors = [1., 1.02, 1.1]
cases = {'state, $m = 50$': dict(m=50), 'state, $m = 5$': dict(m=5),
         'state and parameters': dict(m=50, estimate_parameters=True)}
inflation = {title: {rf'$\Delta t_\mathrm{{obs}} = {dt}\,T_\lambda$':
                     sweep(twin_lorenz63, factors, key='inflation', dt_obs=dt, **options) for dt in (0.1, 0.25)}
             for title, options in cases.items()}
plot_inflation(factors, inflation)
for title, groups in inflation.items():
    print(title.replace('$', ''), {k.replace('$', ''): np.round(np.median(v, axis=1), 2) for k, v in groups.items()})
```

- **State, 50 members:** inflation is not necessary; chaos restores the spread.
- **State, 5 members:** inflation helps when the observations are sparse, because few members sample the covariance poorly.
- **State and parameters:** a large factor degrades the estimate. The parameters have no dynamics, so their inflated spread is neither restored nor bounded.

Chaos makes inflation less necessary for the state, not unnecessary.

## The Kuramoto–Sivashinsky equation <a class="anchor" id="ks"></a>

$$
\frac{\partial u}{\partial t} + u\frac{\partial u}{\partial x} + \frac{\partial^2 u}{\partial x^2} + \frac{\partial^4 u}{\partial x^4} = 0, \qquad x \in [0, 20\pi),
$$

is chaotic with several positive Lyapunov exponents ($T_\lambda \approx 12.5$). The state of the filter is the 33 Fourier coefficients of a 64-point grid, and 32 sensors measure $u$.

```python
plot_ks_fields({rf'$\Delta t_\mathrm{{obs}} = {dt}\,T_\lambda$': twin_ks(dt) for dt in (0.25, 2)})

ks_dt_obs, ks_factors = [0.1, 0.25, 0.5, 1., 2.], [1., 1.05, 1.2]
ks_sampling = sweep(twin_ks, ks_dt_obs, score=None)
ks_inflation = {f'$m = {m}$': sweep(twin_ks, ks_factors, key='inflation', dt_obs=0.1, m=m) for m in (20, 50)}
plot_ks_sweeps(ks_dt_obs, ks_sampling, ks_factors, ks_inflation)
print('median error:', np.round(np.median(ks_sampling[:, :, 0], axis=1), 3))
print({k.replace('$', ''): np.round(np.median(v, axis=1), 3) for k, v in ks_inflation.items()})
```

- **Sampling:** as for the Lorenz-63 system, with a sharper transition: the filter tracks the field up to $\Delta t_\mathrm{obs} = 0.5\,T_\lambda$ and diverges from $T_\lambda$, while its spread stays small.
- **Inflation:** not necessary with 50 members. With 20 members, the filter diverges without inflation and $\rho_I = 1.05$ recovers it: chaos restores the spread only along the directions that the ensemble spans.
